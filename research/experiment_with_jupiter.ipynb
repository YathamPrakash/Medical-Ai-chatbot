```python
from pypdf import PdfReader
from langchain_core.documents import Document

def read_pdf(file_path):
    pages:list[Document]=[]
    reader = PdfReader(file_path)
    text = ""
    for i,page in enumerate(reader.pages):
        text += page.extract_text().strip()
        if(text):
         pages.append(Document(page_content=text,
                              metadata={"page": i}))
        else:
            print(f"No text found on page {i}.")
    if not pages:
       print("No pages found in the PDF.")
    return pages
```

```python
from pathlib import Path
file_path=Path("data/gale-of-encyclopedia-medical.pdf")
print(file_path)
extracteData=read_pdf(file_path=file_path)
```

```python
import re
cleanedData=[]
def clean_extracted_data(extracteData):
     
    for docs in extracteData:
        print(docs)
        text=docs.page_content.replace("\r\n","\n").replace("\r","\n")
        text=re.sub(r"[\t]", ' ',text)
        text=re.sub(r"\n{3,}","\n",text)
        text=re.sub(r" +", " ", text)
        text=text.strip()
        if(text):     
            cleanedData.append(Document(page_content=text,
                                metadata=docs.metadata.copy()))
    if not cleanedData:
            raise ValueError("No cleaned data available.")
    return cleanedData
```
